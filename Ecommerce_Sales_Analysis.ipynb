# Data Cleaning

Before performing analysis, it is important to clean the datasets by checking for duplicate records, missing values, and correcting data types. This ensures accurate analysis and reliable visualizations.

```python
print("Duplicate Records")

print("Customers :", customers.duplicated().sum())
print("Orders :", orders.duplicated().sum())
print("Order Items :", order_items.duplicated().sum())
print("Products :", products.duplicated().sum())
print("Reviews :", reviews.duplicated().sum())
```

```python
customers.drop_duplicates(inplace=True)
orders.drop_duplicates(inplace=True)
order_items.drop_duplicates(inplace=True)
products.drop_duplicates(inplace=True)
reviews.drop_duplicates(inplace=True)

print("Duplicate records removed successfully.")
```

# Data Type Conversion

The order purchase timestamp is converted into datetime format to enable monthly and yearly sales analysis.

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

print("Date converted successfully.")
```

# Merge the Datasets

To perform comprehensive sales analysis, the required datasets are merged using common keys such as order ID, product ID, and customer ID.

```python
merged = orders.merge(
    order_items,
    on="order_id",
    how="inner"
)

merged = merged.merge(
    products,
    on="product_id",
    how="left"
)

merged = merged.merge(
    customers,
    on="customer_id",
    how="left"
)

merged = merged.merge(
    reviews,
    on="order_id",
    how="left"
)

merged = merged.merge(
    translation,
    on="product_category_name",
    how="left"
)

print("Datasets merged successfully.")
```

# Sales Analysis

This section answers the business questions provided in the project brief using the merged dataset.

```python
# Revenue = Product Price + Freight Charges

merged["revenue"] = merged["price"] + merged["freight_value"]

merged[["price","freight_value","revenue"]].head()
```

## Question 1

### Which Product Category Generates the Highest Revenue?

```python
category_revenue = merged.groupby(
    "product_category_name_english"
)["revenue"].sum().sort_values(ascending=False)

category_revenue.head(10)
```

```python
plt.figure(figsize=(12,6))

category_revenue.head(10).plot(
    kind="bar",
    color="steelblue"
)

plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()
```

## Question 2

### Which Month Had the Highest Sales?

```python
merged["Month"] = merged["order_purchase_timestamp"].dt.strftime("%B")

merged["Month_Number"] = merged["order_purchase_timestamp"].dt.month
```

```python
monthly_sales = merged.groupby(
    ["Month_Number","Month"]
)["revenue"].sum().reset_index()

monthly_sales
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["revenue"],
    marker="o"
)

plt.xticks(rotation=45)

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Revenue")

plt.show()
```

## Question 3

### Which State Performs the Best?

```python
state_sales = merged.groupby(
    "customer_state"
)["revenue"].sum().sort_values(ascending=False)

state_sales.head(10)
```

```python
plt.figure(figsize=(12,6))

state_sales.head(10).plot(
    kind="bar",
    color="green"
)

plt.title("Top 10 States by Revenue")

plt.xlabel("State")

plt.ylabel("Revenue")

plt.show()
```

## Question 4

### What is the Average Order Value?

```python
average_order_value = merged.groupby("order_id")["revenue"].sum().mean()

print("Average Order Value (AOV): ₹", round(average_order_value,2))
```

## Question 5

### Customer Review Score Distribution

```python
review_distribution = merged["review_score"].value_counts().sort_index()

review_distribution
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    review_distribution,
    labels=review_distribution.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Customer Review Score Distribution")

plt.show()
```

# Additional Visualizations

```python
plt.figure(figsize=(10,5))

plt.hist(
    merged["revenue"],
    bins=40
)

plt.title("Revenue Distribution")

plt.xlabel("Revenue")

plt.ylabel("Frequency")

plt.show()
```

```python
heatmap_data = merged.pivot_table(
    values="revenue",
    index="product_category_name_english",
    columns="Month_Number",
    aggfunc="sum"
)

plt.figure(figsize=(14,12))

sns.heatmap(
    heatmap_data.head(15),
    cmap="YlGnBu"
)

plt.title("Revenue by Product Category and Month")

plt.show()
```

# Dashboard KPIs

```python
total_revenue = merged["revenue"].sum()

total_orders = merged["order_id"].nunique()

top_category = category_revenue.idxmax()

best_month = monthly_sales.loc[
    monthly_sales["revenue"].idxmax(),
    "Month"
]

print("="*40)

print("TOTAL REVENUE :", round(total_revenue,2))

print("TOTAL ORDERS :", total_orders)

print("TOP CATEGORY :", top_category)

print("BEST MONTH :", best_month)

print("="*40)
```
